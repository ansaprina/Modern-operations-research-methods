```python
import random
import time
```

```python
class MaxCliqueProblem:
    def __init__(self):
        self.neighbour_sets = []
        self.best_clique = []

    def read_graph_file(self, filename):
        with open(filename, "r") as f:
            for line in f:
                line = line.strip()
                if not line or line[0] == 'c':
                    continue
                tokens = line.split()
                if tokens[0] == 'p':
                    vertices = int(tokens[2])
                    self.neighbour_sets = [set() for _ in range(vertices)]
                elif tokens[0] == 'e':
                    u = int(tokens[1]) - 1
                    v = int(tokens[2]) - 1
                    self.neighbour_sets[u].add(v)
                    self.neighbour_sets[v].add(u)

    def find_clique(self, randomization, iterations):
        for _ in range(iterations):
            clique = []
            candidates = list(range(len(self.neighbour_sets)))
            random.shuffle(candidates)
            while candidates:
                limit = min(randomization, len(candidates))
                vertex = random.choice(candidates[:limit])
                clique.append(vertex)
                candidates = [v for v in candidates if v in self.neighbour_sets[vertex]]
                random.shuffle(candidates)
            if len(clique) > len(self.best_clique):
                self.best_clique = clique

    def get_clique(self):
        return self.best_clique

    def check(self):
        if len(self.best_clique) != len(set(self.best_clique)):
            print("Duplicated vertices in the clique")
            return False
        for i in self.best_clique:
            for j in self.best_clique:
                if i != j and j not in self.neighbour_sets[i]:
                    print("Returned subgraph is not a clique")
                    return False
        return True
```

```python
iterations = 10000
randomization = 3

files = [
    "brock200_1.clq", "brock200_2.clq", "brock200_3.clq", "brock200_4.clq", "brock400_1.clq",
    "brock400_2.clq", "brock400_3.clq", "brock400_4.clq", "C125.9.clq", "gen200_p0.9_44.clq",
    "gen200_p0.9_55.clq", "hamming8-4.clq", "johnson16-2-4.clq", "johnson8-2-4.clq", "keller4.clq",
    "MANN_a27.clq", "MANN_a9.clq", "p_hat1000-1.clq", "p_hat1000-2.clq", "p_hat1500-1.clq",
    "p_hat300-3.clq", "p_hat500-3.clq", "san1000.clq", "sanr200_0.9.clq", "sanr400_0.7.clq"
]

with open("clique.csv", "w") as fout:
    fout.write("File; Clique; Time (sec); Clique vertices\n")
    for file in files:
        problem = MaxCliqueProblem()
        try:
            problem.read_graph_file(file)
        except Exception as e:
            print(f"Error reading file {file}: {e}")
            continue
        start_time = time.time()
        problem.find_clique(randomization, iterations)
        elapsed = time.time() - start_time
        if not problem.check():
            print(f"*** WARNING: incorrect clique in {file} ***")
            fout.write("*** WARNING: incorrect clique ***\n")
        clique = problem.get_clique()
        clique_str = ", ".join(str(v + 1) for v in clique)
        fout.write(f"{file}; {len(clique)}; {elapsed:.3f}; {clique_str}\n")
        print(f"{file}, result - {len(clique)}, time - {elapsed:.3f}")
        print(f"Clique vertices: {clique_str}")
```
