```python
import os
import time
```

```python
class ColoringProblem:
    def __init__(self):
        self.neighbour_sets = []
        self.colors = []
        self.maxcolor = 0

    def read_graph_file(self, filename):
        with open(filename, 'r') as fin:
            for line in fin:
                line = line.strip()
                if not line:
                    continue
                if line[0] == 'c':
                    continue
                parts = line.split()
                if parts[0] == 'p':
                    vertices = int(parts[2])
                    self.neighbour_sets = [set() for _ in range(vertices)]
                    self.colors = [0] * vertices
                elif parts[0] == 'e':
                    start = int(parts[1])
                    finish = int(parts[2])
                    self.neighbour_sets[start - 1].add(finish - 1)
                    self.neighbour_sets[finish - 1].add(start - 1)

    def dsatur_graph_coloring(self):
        n = len(self.neighbour_sets)
        colors = [0] * n
        neighbor_colors = [set() for _ in range(n)]
        degrees = [len(neighbours) for neighbours in self.neighbour_sets]
        uncolored = set(range(n))

        while uncolored:
            vertex = max(uncolored, key=lambda v: (len(neighbor_colors[v]), degrees[v]))
            color = 1
            while color in neighbor_colors[vertex]:
                color += 1
            colors[vertex] = color
            for neighbour in self.neighbour_sets[vertex]:
                if neighbour in uncolored:
                    neighbor_colors[neighbour].add(color)
            uncolored.remove(vertex)

        self.colors = colors
        self.maxcolor = max(colors) if colors else 0

    def check(self):
        for i, neighbours in enumerate(self.neighbour_sets):
            if self.colors[i] == 0:
                print(f"Vertex {i+1} is not colored")
                return False
            for neighbour in neighbours:
                if self.colors[neighbour] == self.colors[i]:
                    print(f"Neighbour vertices {i+1} and {neighbour+1} have the same color")
                    return False
        return True

    def get_number_of_colors(self):
        return self.maxcolor

    def get_colors(self):
        return self.colors

    def get_color_classes(self):
        color_classes = {}
        for idx, color in enumerate(self.colors):
            if color not in color_classes:
                color_classes[color] = []
            color_classes[color].append(idx + 1)
        result = []
        for color in sorted(color_classes.keys()):
            vertices = sorted(color_classes[color])
            result.append("{" + ", ".join(map(str, vertices)) + "}")
        return ", ".join(result)
```

```python
files = [
    "myciel3.col", "myciel7.col", "school1.col", "school1_nsh.col", "anna.col",
    "miles1000.col", "miles1500.col", "le450_5a.col", "le450_15b.col",
    "queen11_11.col"
]

with open("color_optimized.csv", "w") as fout:
    header = "Instance\nTime, sec\nColors\nColor classes\n"
    fout.write(header)
    print("Instance")
    print("Time, sec")
    print("Colors")
    print("Color classes")
    for file in files:
        problem = ColoringProblem()
        problem.read_graph_file(file)
        start_time = time.time()
        problem.dsatur_graph_coloring()
        elapsed_time = time.time() - start_time
        if not problem.check():
            fout.write(f"*** WARNING: incorrect coloring in {file} ***\n")
            print(f"*** WARNING: incorrect coloring in {file} ***")
        instance_name = os.path.basename(file)
        number_of_colors = problem.get_number_of_colors()
        color_classes_str = problem.get_color_classes()
        fout.write(f"{instance_name}\n{elapsed_time:.3f}\n{number_of_colors}\n{color_classes_str}\n")
        print(instance_name)
        print(f"{elapsed_time:.3f}")
        print(number_of_colors)
        print(color_classes_str)
```
